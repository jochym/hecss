# Development

Notes for contributors and project setup details (developer-facing).


## VASP runner configuration

`run-calc.sh` is the standard entrypoint for VASP jobs (local or remote). Configuration is read from `run-calc.conf` with this precedence:

1) `./run-calc.conf` in the working directory (recommended per-run)
2) `~/.hecss/run-calc.conf` (user-wide defaults)
3) `~/run-calc.conf` (legacy)
4) a copy adjacent to `example/scripts/run-calc-ssh.sh`

Ensure one of these exists before invoking CLI commands or ASE calculators that call `run-calc.sh`.


## Local VASP pseudopotentials (direnv)

Keep non-redistributable POTCAR trees outside version control:

- Place them under `local/vasp_pp/` in the project root (git-ignored).
- Use `direnv` to scope `VASP_PP_PATH` to this repo. Example `.envrc` (git-ignored):
  ```
  export VASP_PP_PATH=$PWD/local/vasp_pp
  ```
- Enable once with `direnv allow` in the project root.

This keeps `VASP_PP_PATH` out of global shell config and notebooks while making it available to ASE/VASP runs in this project.
